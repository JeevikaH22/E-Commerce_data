```python
df=df.rename(columns={" Customer_Name": "Customer_Name", " Category":"Category"})
```

```python
df = df.drop_duplicates()
```

#### **Value formatting-**

```python
print(df["Category"].unique())
print(df["Payment_Method"].unique())
print(df["Status"].unique())
```

```python
print(df["Product"].unique())
```

```python
df["Category"] = df["Category"].str.strip().str.title()
df["Product"] = df["Product"].str.strip().str.title()
```

```python
df["Category"] = df["Category"].replace({"Electronic": "Electronics"})
```

```python
df["Price"] = df["Price"].astype(str).str.replace("$", "", regex=False).str.strip()
df["Price"] = df["Price"].replace({"four hundred": 400})
```

```python
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
```

```python
df['Price'] = df['Price'].abs()
```

```python
df['Quantity'] = df['Quantity'].astype(str).str.extract(r'(\d+)')[0]
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
```

```python
df['Quantity'] = df['Quantity'].abs()
```

```python
df["Total"] = df["Quantity"] * df ["Price"]
```

```python
df['Category'] = df['Category'].fillna('Unknown')
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='mixed', errors='coerce')
```
